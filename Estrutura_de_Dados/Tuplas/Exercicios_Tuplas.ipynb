# Desafios de Tupla

* 1 - Imutabilidade: A imutabilidade de tuplas não é uma limitação, é uma garantia. Em pipelines de dados, isso significa que um registro passado como tupla entre funções não pode ser corrompido acidentalmente. Entender isso na prática muda como você pensa sobre modelagem de dados.

```python
leitura = ("M07", "2026-01-01 08:00:00", 72.4, 0.6, False) # Campos: máquina, timestamp, temp°, vibração, falha
```

* Desafio 01: Acesse cada campo pelo índice posicional e atribua a variáveis com nomes descritivos. Imprima um relatório formatado: "Máquina M07 | 72.4°C | 0.6 mm/s | Status: OK". Evite usar leitura[2] diretamente na string — use a variável nomeada.

```python
maquina = leitura[0]
timestamp = leitura[1]
temperatura = leitura[2]
vibra = leitura[3]
tem_falha = leitura[4]

if not tem_falha:
    status = "OK"
else:
    status = "FALHA"

print(f"Máquina: {maquina} | Temperatura(C°): {temperatura} | Vibração(mm/s): {vibra} | Status: {status} | Data e Hora: {timestamp}")
```

* Desafio 02: "Modifique" a temperatura para 80.0 criando uma nova tupla: use concatenação de slices. Atribua a leitura_corrigida. Esse padrão é como você trata correções em registros imutáveis sem mutação.

```python
leitura_corrigida = leitura[:2] + (80.0,) + leitura[3:]
leitura_corrigida
```

* Desafio 03: Use slicing para extrair apenas os campos numéricos (temperatura e vibração) em uma sub-tupla. Use len() para verificar o tamanho e o operador in para checar se "M07" está na tupla.

```python
leitura_selecionada = leitura[2:4]
print(len(leitura_selecionada))
if "M07" in leitura_selecionada:
    print("M07 está na tupla")
else:
    print("M07 não está na tupla")
```
