#  Dicionário

* O dicionário ou dict armazena
uma coleção de pares chave-valor, em que a chave e o valor são objetos
Python. Cada chave é associada a um valor para que um valor possa ser
convenientemente recuperado, inserido, modificado ou excluído dada uma
chave específica. Uma abordagem para a criação de um dicionário seria com
o uso de chaves ({}) e dois pontos para separar chaves e valores:

```python
dict_vazio = {}
d1 = {"a": "some value", "b": [1, 2, 3, 4]}
d1
```

* Você pode acessar, inserir ou definir elementos usando a mesma sintaxe do acesso a elementos de uma lista ou tupla:

```python
d1[7] = "an integer"
d1
```

```python
d1["b"]
```

* É possível verificar se um dicionário contém uma chave com o uso da mesma sintaxe utilizada para verificar se uma lista ou tupla contém um valor:

```python
"b" in d1
```

```python
"D" in d1
```

* Podemos excluir valores usando a palavra-chave del ou o método pop (que simultaneamente retorna o valor e exclui a chave):

```python
d1[5] = "some value"
d1["dummy"] = "some value"
d1
```

```python
# Excluindo chaves:
del d1[5]
ret = d1.pop("dummy")
ret
```

```python
d1
```

* Se você precisar iterar tanto pelas chaves quanto pelos valores, pode usar o método items para percorrê-los como tuplas de dois elementos:

```python
list(d1.items())
```

* Os métodos keys e values fornecem iteradores para as chaves e os valores do dicionário, respectivamente. A ordem das chaves vai depender da ordem de sua inserção e essas funções exibem as chaves e os valores na mesma ordem:

```python
list(d1.keys())
```

```python
list(d1.values())
```

* Você pode mesclar dois dicionários usando o método update:

```python
d1.update({"b": "foo", "c": 12})
d1
```

* O método update altera os dicionários diretamente no local, logo, qualquer chave existente nos dados passados para update terá seu valor anterior descartado.

* Criação de dicionários a partir de sequências:

* É comum, ocasionalmente, termos duas sequências e querermos emparelhá-las elemento a elemento em um dicionário. Como uma primeira tentativa, você poderia escrever um código como este:

```python
mapping = {}
key_list = ["A", "B", "C", "D"]
value_list = [10, 2, 4, 14]
for key, value in zip(key_list, value_list):
    mapping[key] = value
mapping
```

* Valor Padrão:

* O conceito de Valores Padrão serve para evitar o erro KeyError quando você tenta acessar uma chave que não existe no dicionário.
Em vez de o programa travar, você define o que deve ser devolvido ou criado automaticamente.
Existem três formas principais de lidar com isso, dependendo do que você precisa fazer:

1. Apenas leitura: .get()
O método d.get(chave, valor_padrao) busca a chave. Se ela não existir, ele devolve o valor_padrao (ou None por padrão).
Importante: Ele não altera o dicionário; apenas devolve o valor para você usar naquela linha.

2. Inserção e Atualização: .setdefault()
Este método é o "atalho" para quando você quer garantir que a chave exista antes de usá-la.
Como funciona: Se a chave não existir, o Python a insere no dicionário com o valor padrão fornecido e retorna esse valor.
Vantagem: Permite atualizar valores mutáveis (como listas) em uma única busca, como no exemplo by_letter.setdefault(letra, []).append(palavra).

3. Automatização: defaultdict
Disponível no módulo collections, essa classe automatiza a criação de valores para chaves ausentes sempre que você acessa d[chave].
Fábrica de valores: Você define uma "fábrica" (como list, int ou dict) ao criar o dicionário. Se a chave não for encontrada, a fábrica é chamada, o novo item é inserido e devolvido pronto para uso.
Diferença: Ele só dispara o valor padrão no acesso direto via colchetes [], não via .get().
Resumindo: use .get() para consultas seguras, .setdefault() para inserções rápidas em dicionários comuns, e defaultdict quando souber que seu dicionário sempre precisará de um tipo específico de valor para chaves novas.

```python
from collections import defaultdict

words = ["apple", "bat", "bar", "atom", "book"]

# 1. Abordagem Manual (if-else)
# É necessário verificar se a chave existe antes de adicionar itens [2].
by_letter_manual = {}
for word in words:
    letter = word
    if letter not in by_letter_manual:
        by_letter_manual[letter] = [word]
    else:
        by_letter_manual[letter].append(word)

# 2. Usando .setdefault()
# Simplifica o fluxo: se a chave não existir, ele a cria com uma lista vazia [] e já retorna essa lista para o .append() [2, 3].
by_letter_sd = {}
for word in words:
    letter = word
    by_letter_sd.setdefault(letter, []).append(word)

# 3. Usando defaultdict (A mais elegante)
# Você define que todo novo item será automaticamente uma lista (list). O acesso direto cria o objeto para você [4, 5].
by_letter_dd = defaultdict(list)
for word in words:
    by_letter_dd[word].append(word)

print(dict(by_letter_dd)) 
# Resultado: {'a': ['apple', 'atom'], 'b': ['bat', 'bar', 'book']}
```
