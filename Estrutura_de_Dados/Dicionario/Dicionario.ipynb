#  Dicionário

* O dicionário ou dict armazena
uma coleção de pares chave-valor, em que a chave e o valor são objetos
Python. Cada chave é associada a um valor para que um valor possa ser
convenientemente recuperado, inserido, modificado ou excluído dada uma
chave específica. Uma abordagem para a criação de um dicionário seria com
o uso de chaves ({}) e dois pontos para separar chaves e valores:

```python
dict_vazio = {}
d1 = {"a": "some value", "b": [1, 2, 3, 4]}
d1
```

* Você pode acessar, inserir ou definir elementos usando a mesma sintaxe do acesso a elementos de uma lista ou tupla:

```python
d1[7] = "an integer"
d1
```
