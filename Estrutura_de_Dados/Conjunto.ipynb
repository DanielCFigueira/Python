# Conjunto(set)

* Um conjunto (set) é uma coleção não ordenada de objetos únicos e imutáveis. Diferente das listas, ele não permite duplicatas: cada item aparece apenas uma vez, independentemente de quantas vezes seja adicionado.
Ele pode ser criado de duas maneiras: com a função set ou por meio de um conjunto literal com chaves:

```python
set([2, 2, 1, 3])
```

```python
{2, 2, 2, 1, 3, 3}
```

* Operações Matemáticas

* Os conjuntos suportam operações matemáticas de conjunto, como a união, interseção, diferença e diferença simétrica. Considere estes dois exemplos de conjuntos:

```python
a = {1, 2, 3, 4, 5}
b = set([3, 4, 5, 6, 7, 8])
```

* A união desses dois conjuntos fornecerá o conjunto de elementos distintos que ocorrem em cada conjunto. Isso pode ser obtido com o método union ou com o operador binário |:

```python
a.union(b)
```

```python
a | b
```

* A interseção conterá os elementos que estiverem ocorrendo nos dois conjuntos. O operador & ou o método intersection podem ser usados:

```python
a.intersection(b)
```

```python
a & b
```

* Todas as operações lógicas de conjuntos têm contrapartidas diretamente no local, o que permite substituir o conteúdo do conjunto do lado esquerdo da operação pelo resultado. Para conjuntos muito grandes, isso pode ser mais eficiente:

```python
c = a.copy()
c
```

```python
c |= b
c 
```

```python
d = a.copy()
d
```

```python
d &= b
d
```

* Como as chaves dos dicionários, geralmente os elementos dos conjuntos
precisam ser imutáveis e hashable (ou seja, se você chamar hash em um valor
isso não lançará uma exceção). Para armazenar elementos de tipo lista (ou
outras sequências mutáveis) em um conjunto, você pode convertê-los em
tuplas:

```python
my_data = [1, 2, 3, 4]
my_set = {tuple(my_data)}
my_set
```

* Você também pode verificar se um conjunto é um subconjunto (se está
contido em) ou um superconjunto (se contém todos os elementos) de outro
conjunto:

```python
a_set = {1, 2, 3, 4, 5}
a_set
```

```python
{1,2,3}.issubset(a_set) # 1, 2 e 3 estão contidos em a_set.
```

```python
{1, 2, 6}.issubset(a_set) # 1, 2 e 6 não estão contidos em a_set.
```

```python
a_set.issuperset({1, 2, 3}) # a_set contém todos os elementos.
```

```python
a_set.issuperset({2, 6, 9}) # a_set não contém todos os elementos.
```

* Os conjuntos serão iguais se, e somente se, os seus conteúdos forem iguais:

```python
{1, 2, 3} == {3, 2, 1}
```

```python
{4, 5, 6} == {6, 5, 0}
```
