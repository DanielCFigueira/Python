# Funções de Sequência Internas

* Enumerate:
Na iteração por uma sequência, podemos querer registrar o índice do item atual, para isso o Python tem uma função interna chamada enumerate.
Esta função é utilizada quando você precisa iterar por uma sequência e, ao mesmo tempo, manter o controle do índice do item atual. Ela retorna um gerador que produz tuplas no formato (índice, valor).

```python
nomes = ["Daniel", "Max", "Magnus"]
for i, nome in enumerate(nomes):
    print(f"{i}: {nome}")
```

* Sorted: A função sorted retorna uma nova lista ordenada a partir dos elementos de qualquer sequência. Ao contrário do método .sort(), que modifica a lista original "no lugar", a função sorted() cria e retorna uma nova lista ordenada a partir de qualquer objeto iterável.
Dica: Ela aceita argumentos como reverse=True para ordem decrescente e key para definir um critério de ordenação personalizado (como o tamanho de strings).

```python
# 1. Uso básico: Ordena uma lista numérica ou uma string (neste caso, retorna uma lista de caracteres ordenados)

numeros = [5, 2, 6, 0, 3, 1, 4, 7, 10, 8, 9]
print(sorted(numeros))

print(sorted("python"))
```

```python
# 2. Ordem Inversa:

carros = ['bmw', 'audi', 'toyota', 'subaru']
print(sorted(carros, reverse=True))
```

```python
# 3. Ordenação com Chave Personalizada (key): Este é um recurso poderoso que permite definir um critério. 
# Por exemplo, ordenar strings pelo tamanho (len) em vez da ordem alfabética

frutas = ["açaí", "uva", "maça", "banana"]
print(sorted(frutas, key=len))
```

```python
# 4. Ordenando Dicionários: Ao usar sorted em um dicionário, ele retorna uma lista ordenada das chaves por padrão

precos = {"maça": 5.50, "banana": 3.00, "uva": 8.00}
print(sorted(precos))
```

* Zip: zip “gera pares” com os elementos de várias listas, tuplas ou outras sequências para criar uma lista de tuplas. A função zip pode receber um número arbitrário de sequências, e o número de elementos que ela produzirá é determinado pela sequência mais curta.

```python
seq1 = ["foo", "bar", "baz"]
seq2 = ["um", "dois", "três"]
zipped = zip(seq1, seq2)
list(zipped)
```

```python
seq3 = [False, True]
list(zip(seq1, seq2, seq3))
```

```python
# zip com iteração em Laços for:

produtos = ["maça", "uva", "banana"]
precos = [5.50, 7.30, 10.00]

for produto, preco in zip(produtos, precos):
    print(f"O preço de {produto} é R${preco:.2f}")
```

* Reversed: reversed itera pelos elementos de uma sequência na ordem inversa. 

```python
list(reversed(range(11)))
```

* Lembre-se de que reversed é um gerador (o que será discutido com mais
detalhes posteriormente), logo, ela não cria a sequência inversa até que esta
seja materializada (por exemplo, com list ou um loop for).
