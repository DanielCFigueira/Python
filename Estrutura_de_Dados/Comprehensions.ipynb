# List, set e dictionary comprehensions

* As list comprehensions (compreensões de lista) são um recurso que cria novas coleções em Python (listas, conjuntos ou dicionários) a partir de um iterável, sem a necessidade de escrever várias linhas de código com laços for tradicionais.

* Sintaxes:

1. List Comprehensions(Listcomps): [expressão for item in coleção if condição]

```python
quadrados = [X**2 for X in range(1,11)]
quadrados
```

* Filtragem: Você pode adicionar um if no final para incluir apenas os itens que atendam a um critério. Por
exemplo, dada uma lista de strings, poderíamos excluir por filtragem strings
de tamanho 2 ou menores e converter as strings restantes para letras
maiúsculas desta forma:

```python
strings = ["a", "as", "bat", "car", "dove", "python"]
[x.upper() for x in strings if len(x) > 2]
```

* 2. Set e Dictionary Comprehensions:


* Uso prático do set: Obter o tamanho único de várias strings {expressão for valor in coleção if condição}:

```python
tamanhos = {len(x) for x in ["apple", "bat", "apple"]}
tamanhos
```

* Uso prático do dict: {chave_expr: valor_expr for item in coleção if condição}

```python
strings = ['a', 'as', 'bat', 'car']
loc_mapping = {val: i for i, val in enumerate(strings)}
loc_mapping
```

* Como as list comprehensions, as set e dictionary comprehensions são em
grande parte conveniências, mas também podem tornar o código mais fácil
tanto de ler quanto de escrever. Considere a lista de strings anterior.
Suponhamos que quiséssemos um conjunto contendo apenas os tamanhos das
strings da coleção; poderíamos obter esse resultado facilmente usando uma
set comprehension:

```python
Strings = ["a", "as", "bat", "car", "dove", "python"]
unique_lengths = {len(x) for x in Strings}
unique_lengths
```

* Também poderíamos expressar isso de maneira mais funcional usando a
função map, que será apresentada em breve:

```python
set(map(len, Strings))
```

* Como um exemplo simples de uma dictionary comprehension, poderíamos
criar um mapa de consulta dessas strings para descobrir seus locais na lista:

```python
loc_mapping_dict = {value: index for index, value in enumerate(Strings)}
loc_mapping_dict
```

* List Comprehensions Aninhadas: As List Comprehensions Aninhadas permitem encapsular operações que normalmente exigiriam vários laços for em uma única expressão concisa. Elas são muito úteis para "achatar" estruturas (como converter uma lista de listas em uma lista simples) ou para realizar produtos cartesianos.

- Ordem dos Laços: As cláusulas for seguem a mesma ordem em que seriam escritas em laços tradicionais: o laço mais externo vem primeiro, seguido pelos internos.
- Sintaxe Básica: [expressão for item_externo in colecao_externa for item_interno in item_externo].
Filtragem: Você pode adicionar uma condição if no final para filtrar os elementos resultantes.

```python
# 1. Linearização (Flattening): Transforma uma lista de listas ou uma lista de tuplas em uma lista simples e "achatada".

some_tuples = [(1, 2, 3), (4, 5, 6), (7, 8, 9)]
flattend = [x for tup in some_tuples for x in tup]
flattend
```

```python
# 2. Filtragem em Estruturas Aninhadas: Permite percorrer sublistas e extrair apenas os itens que atendem a um critério.

all_data = [
    ["John", "Emily", "Michael", "Mary", "Steven"],
    ["Maria", "Juan", "Javier", "Natalia", "Pilar"]
]

# Busca nomes que contenham dois ou mais a's:
result = [name for names in all_data for name in names if name.count("a") >=2]
result
```

```python
# 3. Produto Cartesiano: Gera todas as combinações possíveis entre os elementos de dois ou mais iteráveis.

colors = ['black', 'red']
sizes = ['S', 'M', 'L']

# Cria uma lista de tuplas combinando todas as cores e tamanhos:

tshirts = [(color, size) for color in colors for size in sizes]
tshirts
```

```python
# 4. Processamento de Matrizes: Você pode realizar cálculos entre matrizes (listas de listas). 
# Note que, para manter a estrutura de matriz no resultado, usa-se uma comprehension dentro de outra:

# M e N são matrizes 3x3. Multiplica elementos na mesma posição.

M = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

N = [
    [9, 8, 7],
    [6, 5, 4],
    [3, 2, 1]
]

res = [[M[row][col] * N[row][col] for col in range(3)] for row in range(3)]
res
```
