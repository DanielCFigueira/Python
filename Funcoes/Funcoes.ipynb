# Funções: 

* As funções são o principal e mais importante método de organização e
reutilização de código em Python. Como regra geral, se você detectar que
precisará repetir o mesmo código ou um código semelhante mais de uma vez,
pode ser melhor escrever uma função reutilizável. As funções também podem
ajudar a tornar seu código mais legível fornecendo um nome para um grupo
de instruções Python.

* As funções são declaradas com a palavra-chave def. Uma função contém um
bloco de código e opcionalmente pode usar a palavra-chave return:


```python
def my_function(x, y):
    return x + y
```

* Quando uma linha com return é alcançada, o valor ou a expressão seguinte a
return é enviado para o contexto no qual a função foi chamada; por exemplo:

```python
my_function(1,  2)
```

```python
result = my_function(7, 5)
result
```

* Não é um problema haver várias instruções return. Se o Python alcançar o fim
de uma função sem encontrar uma instrução return, None será retornado
automaticamente. Por exemplo:

```python
def function_without_return(x):
    print(x)
result = function_without_return("Hello!")
print(result)
```

* Cada função pode ter argumentos posicionais e argumentos nomeados. Os
argumentos nomeados são mais usados para especificar valores padrão ou
argumentos opcionais. É importante diferenciar a forma como a função é chamada para a forma como ela foi definida.

* -> Eles são definidos na criação da função (def func(x=10):) para tornar um parâmetro opcional. Se você não enviar um valor para o parâmetro x, o Python usará automaticamente o valor 10(valor definido por padrão):

```python
# Definição da função com um valor padrão(Default):
def descrever_animal(nome, especie="cachorro"):
    print(f"Eu tenho um(a) {especie} chamado {nome.title()}.")

# 1. Usando o Valor Padrão (Default):
# O parâmetro 'especie' é opcional. Se omitido, o Python usa o valor 'cachorro'.
descrever_animal("Rex")
```

* -> Argumentos Posicionais: São identificados pela sua ordem (da esquerda para a direita). O primeiro valor que você passa vai para o primeiro parâmetro, e assim por diante:

```python
# 2. Chamada com argumentos posicionais:
# A ordem importa: o primeiro valor passado vai para o primeiro parâmetro definido na criação da função(no caso, nome), 
# já o segundo valor vai para o segundo parâmetro e assim por diante.

descrever_animal("Willie", "Hamester")
descrever_animal("gato", "Rossi")
```

* -> Argumentos Nomeados (ou Keywords): São identificados pelo nome do parâmetro na hora da chamada (ex: func(idade=25)). Isso permite passar os valores em qualquer ordem.

```python
# 3. Chamada com argumentos nomeados(keywwords):
# A ordem não importa: os nomes garantem que os valores cheguem aos parâmetros certos:

descrever_animal(especie="gato", nome="Rossi")
```

# Namespaces, Escopo e Funções Locais:

* As funções podem acessar variáveis criadas tanto dentro da função quanto
fora dela em escopos superiores (ou até mesmo globais). Um nome
alternativo e mais descritivo para o escopo de uma variável em Python é
namespace. Qualquer variável que receba uma atribuição dentro de uma
função por padrão estará associada ao namespace local. O namespace local é
criado quando a função é chamada e é imediatamente preenchido com os
argumentos da função. Depois que a função termina, o namespace local é
destruído. Considere a função a seguir:

```python
def func():
    a = []
    for i in range(5):
        a.append(i)
    return a
teste = func()
print(teste)
```

* Quando func() é chamada, a lista vazia a é criada, cinco elementos são
acrescidos, e depois a é destruída quando a função termina. Suponhamos que
em vez disso tivéssemos declarado a como descrito a seguir:

```python
a = []
def func2():
    for i in range(5):
        a.append(i)
    return a
teste2 = func2()
print(teste2)
```

* Cada chamada a func modificará a lista a:

```python
teste2 = func2()
print(teste2)
```

# Retorno de Vários Valores: 
* Em Python, uma função pode retornar vários valores de uma maneira muito simples e elegante. Embora pareça que ela está devolvendo vários objetos separados, tecnicamente ela retorna um único objeto: uma tupla. O que está ocorrendo aqui é que na verdade a
função está retornando apenas um objeto, uma tupla, que é então
desempacotada gerando as variáveis do resultado.

```python
def calcular_estatistica(numeros):
    soma = sum(numeros)
    media = soma / len(numeros)

    return soma, media # Retorna uma tupla implicitamente

total, avg = calcular_estatistica([2, 4, 6, 8, 9, 10, 12, 14])
total, avg
```

* Uma alternativa seria retornar um dicionário, facilitando a informação de ser identificado:

```python
def func_dict():
    a = 5
    b = 6
    c = 7

    return {"a": a, "b": b, "c": c}

teste3 = func_dict()
print(teste3)
```

# Funções são objetos: 
* Já que as funções Python são objetos, muitas estruturas podem ser facilmente expressas. Mas o que isso que dizer ?



- Você pode guardar uma função dentro de uma variável e chamá-la por meio desse novo nome:

```python
def somar_um(numero):
    return numero + 1 

# Atribuindo a função à variável 'operacao' (sem usar parênteses):
operacao = somar_um

# Chamando a função através da variável:
resultado = operacao(5)
print(resultado)
```

- Ser passada como argumento: Você pode enviar uma função para dentro de outra função como se fosse um dado qualquer. Você envia uma função para dentro de outra para que ela seja executada lá dentro (Higher-Order Function).

```python
def gritar(texto):
    return texto.upper() + "!!!"

def processar_texto(funcao_formatadora, texto_original):
    # Executar a função que foi recebida por parâmetro:
    return funcao_formatadora(texto_original)

# Passando a função 'gritar' como se fosse um dado comum:
print(processar_texto(gritar, "olá mundo"))
```

* - Ser retornada por outra função: Uma função pode criar e devolver outra função como resultado de sua execução.

```python
def criar_multiplicador(fator):
    def multiplicador(numero):
        return numero * fator
    return multiplicador  # Retorna a função interna sem executá-la

# Gerando funções personalizadas:
duplicar = criar_multiplicador(2)
triplicar = criar_multiplicador(3)

print(duplicar(10))
print(triplicar(20))
```

- Ser armazenada em estruturas de dados: Você pode colocar funções dentro de listas, tuplas ou dicionários.

```python
def ligar(): return "Dispositivo ligado"
def desligar(): return "Dispositivo desligado"

# Armazenando funções em um dicionário de comandos:
comandos = {
    "on": ligar,
    "off": desligar
}

# Buscando e executando a função dinamicamente a partir da chave do dicionário:
print(comandos["on"]())
print(comandos["off"]())
```

- Possuir atributos e métodos: Assim como qualquer objeto, uma função tem propriedades internas (como __name__ ou __doc__) e você pode até criar novos atributos nela.

```python
def calcular_area(base, altura):
    """Calcula a área de um retângulo."""
    return base * altura

# 1. Acessando atributos nativos (metadados):
print(calcular_area.__name__)
print(calcular_area.__doc__)

# 2. Criando um atributo personalizado estático na própria função:
calcular_area.versao = "1.0.4"
print(calcular_area.versao)
```
